### Описание проекта

__Общая информация:__

Конструкторский центр разрабатывает инженерные решения для нестандартных ремонтов и модификаций гражданских самолётов, а также консультирует авиакомпании по вопросам поддержания уровня исправности воздушных судов (далее - ВС), необходимого для безопасных полётов.

В работе конструкторского центра (далее - КЦ) используются описывающие проблему запросы авиакомпаний-заказчиков, а на выходе предоставляется инженерное решение (ремонт, модификация, консультация или исследование), разработанное с учётом срочности запроса.

В зависимости от расписания полётов, срочность запроса имеет три категории:

- AOG (Aircraft-On-the-Ground) - требуется срочное решение, т.к. ВС уже должно вылетать, но не может из-за возникшей проблемы;

- Critical - нужно решить проблему в течение 3-5 дней;

- Routine - для разработки решения есть 1-2 недели.

Данный проект разработан для получения общего обзора результатов работы КЦ за рассматриваемый период времени, а также для выявления закономерностей, влияющих на отмену заказов, поскольку отмена заказа приводит к финансовым издержкам. Кроме того, одной из задач проекта является анализ трудозатрат при разработке инженерных решений.

__Цель проекта:__

- Представить общие статистические данные по полученным за рассматриваемый период запросам;

- Выявить закономерности, влияющие на количество отмененных запросов;

- Проанализировать распределение времени, затраченного на разработку решений.

__Данные:__

Для проекта использованы данные о поступивших в КЦ запросах за период с 01 апреля 2023 года по 29 июля 2025 года, собранные в файле Requests_29.07.2025.xlsx.

__Задачи:__

В рамках данного проекта необходимо:

- Провести предварительную обработку данных и подготовить их к дальнейшему анализу;

- Провести анонимизацию персональных данных специалистов, регистрационных номеров ВС и названий компаний-заказчиков;

- Проанализировать основные признаки запроса: тип запроса и его срочность, тип ВС и затронутая система;

- Выявить основные параметры отменённых запросов: типы запросов и их срочность, тип ВС и затронутая система, заказчик и трудозатраты на предварительный анализ;

- Проанализировать распределение трудозатрат на разработку инженерных решений: оценить фактически затраченное время на обработку запросов разных типов, а также сравнить планируемые и фактические трудозатраты;

- Подготовить датасет для загрузки в Tableau/DataLens.

### 4.1 Оценка объёма отменённых проектов

Работа над запросом может быть отменена как заказчиком, так и КЦ на основании предварительного анализа запроса.

Заказчик может отменить свой запрос, если по результатам предварительного анализа, проведённого специалистами КЦ, заказчика не устраивает оценка итоговой стоимости или сроков исполнения заказа.

КЦ может отменить дальнейшую работу над запросом, если в результате предварительной оценки выявляется отсутствие необходимой для реализации проекта информации.

В обоих случаях предварительный анализ проводится за счёт КЦ.

```python
# оценка объёма запросов, имеющих статус "Завершён" и "Отменён"
completed_projects_df = data[data['status'] == 'Завершен']
cancelled_projects_df = data[data['status'] == 'Отменен']

# 1. Оценка количества уникальных завершённых проектов
unique_completed_projects = completed_projects_df['project_id'].nunique()

# 2. Оценка доли уникальных завешённых проектов от всех проектов
total_unique_projects = data['project_id'].nunique()
percentage_completed_projects = (unique_completed_projects / total_unique_projects) * 100

# 3. Оценка суммы фактических трудозатрат на завершённые проекты
total_hours_fact_completed = completed_projects_df['hours_fact'].sum()

# 4. Оценка средних трудозатрат на завершённый проект
average_hours_fact_completed = completed_projects_df['hours_fact'].mean()

# 5. Оценка количества уникальных отменённых проектов
unique_cancelled_projects = cancelled_projects_df['project_id'].nunique()

# 6. Оценка доли уникальных отменённых проектов от всех проектов
percentage_cancelled_projects = (unique_cancelled_projects / total_unique_projects) * 100

# 7. Оценка суммы фактических трудозатрат на отменённые проекты
total_hours_fact_cancelled = cancelled_projects_df['hours_fact'].sum()

# 8. Оценка средних трудозатрат на отменённый проект
average_hours_fact_cancelled = total_hours_fact_cancelled / unique_cancelled_projects

print("\n--- Сравнение завершённых и отменённых проектов ---")

print("\nЗавершённые проекты:")
print(f"  - Количество уникальных завершённых проектов: {unique_completed_projects}")
print(f"  - Процент от общего числа проектов: {percentage_completed_projects:.2f}%")
print(f"  - Общие фактические часы: {total_hours_fact_completed:.2f}")
print(f"  - Средние фактические часы на проект: {average_hours_fact_completed:.2f}")

print("\nОтменённые проекты:")
print(f"  - Количество уникальных отменённых проектов: {unique_cancelled_projects}")
print(f"  - Процент от общего числа проектов: {percentage_cancelled_projects:.2f}%")
print(f"  - Общие фактические часы: {total_hours_fact_cancelled:.2f}")
print(f"  - Средние фактические часы на проект: {average_hours_fact_cancelled:.2f}")
```

__Итого:__ Отменённые проекты составляют более четверти от общего числа проектов. На отменённые проекты в сумме было потрачено 7844 часа (в среднем 9.4 часа на каждый отменённый проект).

### 5.1 Анализ фактических трудозатрат

```python
# средние и суммарные трудозатраты по типам запросов
average_hours_fact_by_req_type = data.groupby('req_type')['hours_fact'].mean().reset_index()
average_hours_fact_by_req_type.columns = ['Тип запроса', 'Среднее фактическое время (часы)']

summ_hours_fact_by_req_type = data.groupby('req_type')['hours_fact'].sum().reset_index()
summ_hours_fact_by_req_type.columns = ['Тип запроса', 'Общее время (часы)']

hours_fact_by_req_type = pd.merge(average_hours_fact_by_req_type, summ_hours_fact_by_req_type, on='Тип запроса', how='left')

hours_fact_by_req_type
```

__Итого:__ Среднее фактическое время, затраченное на один запрос, сильно меняется в зависимости от типа запроса: самыми трудозатратными являются модификации, наименьшее среднее время у консультаций. При этом модификации в три раза трудозатратнее, чем ремонты и в 5 раз - чем консультации. Ремонтов почти в 3 раза больше, чем модификаций, но общие трудозатраты на оба типа запросов сопоставимы (68114 часов у ремонтов против 60737 часов у модификаций.

```python
# общий 95-й персентиль для всех типов запросов
overall_percentile_95_hours = data['hours_fact'].quantile(0.95)

print(f"Общий 95-й перцентиль для фактических трудозатрат (hours_fact) по всем типам запросов: {overall_percentile_95_hours:.2f} часов")
```

```python
# 95-й персентиль для запросов по типам
percentile_95_by_req_type = data.groupby('req_type')['hours_fact'].quantile(0.95).reset_index()
percentile_95_by_req_type.columns = ['Тип запроса', '95-й перцентиль фактических часов']
percentile_95_by_req_type
```

__Итого:__ Поскольку имеются экстремально трудозатратные проекты по модификациям (более 4000 часов), правильно будет сравнивать между собой 95-персентили по типам запросов, т.е. объём фактических часов, в который укладывается 95% всех запросов одного типа.

При сравнении 95-персентилей разных типов запросов выяснилось, что самыми трудозатратными становятся исследования - 95% проектов укладываются в 327,5 часов. за ними следуют модификации - 265 часов, 95% ремонтов укладываются в 80 часов, а консультации - в 73 часа.

```python
# распределние hours_fact по типам запросов
completed_request_filtered_data = data[(data['hours_fact'] <= 330) & (data['status'] == 'Завершен')] #завершённые проекты с hours_fact < 330 часов

plt.figure(figsize=(12, 7))
sns.boxplot(x='req_type', y='hours_fact', data=completed_request_filtered_data, hue='req_type', palette='viridis', legend=False)
plt.title('Распределение фактических трудозатрат по типам запросов (hours_fact < 330 часов)', fontsize=18)
plt.xlabel('Тип запроса', fontsize=16)
plt.ylabel('Фактические часы', fontsize=16)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

Кроме того, модификации имеют гораздо бОльший разброс затраченных часов, чем ремонты, что может свидетельствовать о более высокой степени унификации или стандартизации проектов типа ремонт, где отклонение от 40 часов - скорее выброс, чем норма.

```python
# сравнение средних трудозатрат и 95-персентиля по типам запросов
comparison_df = pd.merge(average_hours_fact_by_req_type, percentile_95_by_req_type, on='Тип запроса') #объединение df со средними трудозатратами по типам запросов и df с 95-м персентилем

comparison_melted = comparison_df.melt(id_vars='Тип запроса', var_name='Метрика', value_name='Часы') #разделение метрик для сравнения

plt.figure(figsize=(12, 7))
ax = sns.barplot(x='Тип запроса', y='Часы', hue='Метрика', data=comparison_melted, palette='viridis')#palette='muted'

plt.title('Сравнение среднего и 95-го перцентиля фактических трудозатрат по типам запросов', fontsize=18)
plt.xlabel('Тип запроса', fontsize=16)
plt.ylabel('Часы', fontsize=16)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title='Метрика', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)

for p in ax.patches:
    ax.annotate(f'{p.get_height():.2f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', fontsize=10, color='black', xytext=(0, 5),
                textcoords='offset points')

plt.tight_layout()
plt.show()
```
